```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
from google.colab import files
import matplotlib.pyplot as plt

# Code added to define 'img' from uploaded files, addressing NameError in subsequent cells.
# This assumes 'uploaded' is a dictionary available in the global scope from a previous execution of `files.upload()`.
try:
    if 'images.jpg' in uploaded:
        img_bytes = uploaded['images.jpg']
        # Convert bytes to numpy array
        img_np = np.frombuffer(img_bytes, np.uint8)
        # Decode image
        img = cv2.imdecode(img_np, cv2.IMREAD_COLOR)
        print("Image 'images.jpg' loaded successfully into 'img'.")
    else:
        print("Warning: 'images.jpg' not found in 'uploaded' dictionary. 'img' not loaded.")
except NameError:
    print("Warning: 'uploaded' variable not found. Please ensure `files.upload()` has been executed in a prior cell.")
except Exception as e:
    print(f"An error occurred while loading the image: {e}")
```

```python

uploaded = files.upload()
```

```python

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python

plt.title("Grayscale Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.hist(gray.ravel(), 256, [0,256])
plt.show()
```

```python
colors = ('b', 'g', 'r')

for i, col in enumerate(colors):
    hist = cv2.calcHist([img], [i], None, [256], [0,256])
    plt.plot(hist, color=col)

plt.title("RGB Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.show()
```

```python
equalized = cv2.equalizeHist(gray)
cv2_imshow(equalized)
```

```python


plt.title("Equalized Histogram")
plt.hist(equalized.ravel(), 256, [0,256])
plt.show()
```
